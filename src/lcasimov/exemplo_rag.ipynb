```python
from langchain_core.documents import Document

docs = [
    Document(page_content="A política de férias da empresa permite 30 dias por ano."),
    Document(page_content="O time de engenharia usa LangChain para construir agentes de IA."),
    Document(page_content="Os benefícios incluem plano de saúde e vale alimentação."),
    Document(page_content="O expediente padrão é de segunda a sexta, das 9h às 18h."),
]
```

```python
from openai import embeddings
# from langchain_openai import OpenAIEmbeddings
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore

# embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")


vectorstore = InMemoryVectorStore.from_documents(
    documents = docs,
    embedding=embeddings,
)
```

```python
retriever = vectorstore.as_retriever(search_kwargs={"k":2})
```

```python
perguntas = [
    "Quantos dias de férias eu tenho no ano?",
    "Quais benefícios a empresa oferece?",
    "Qual ferramentas são utilizadas pela equipe de engenharia?",
    "Qual o horário de expediente?"
]
```

```python
for pergunta in perguntas:
    print("=" * 60)
    print("PERGUNTA :", pergunta)
    resultados = retriever.invoke(pergunta)
    for i, doc in enumerate(resultados, start=1):
        print(f"Resultado {i}: {doc.page_content}")
```
